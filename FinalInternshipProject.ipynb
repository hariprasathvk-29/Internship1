Calculate Total Marks

```python
import pandas as pd
df = pd.read_csv("Data.csv")

subjects = ['Math', 'Physics', 'Chemistry', 'English']
df['Total'] = df[subjects].sum(axis=1)

df[['Student_Name', 'Total']].head()
```

Calculate Average Mark

```python
df['Average']=df[subjects].mean(axis=1)
df[['Student_Name', 'Average']].head(7)
```

```python
df.columns
```

Assingning Grades

```python
def grade(avg):
  if avg >= 90:
    return 'A+'
  elif avg >= 80:
    return 'A'
  elif avg >= 70:
    return 'B'
  elif avg >= 60:
    return 'C'
  elif avg >=50:
    return 'D'
  else:
    return 'F'

df['Grade']=df['Average'].apply(grade)
df[['Student_Name', 'Average', 'Grade']].tail(10)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Data.csv")
subjects = ['Math', 'Physics', 'Chemistry', 'English']

subject_avg = df[subjects].mean()

subject_avg.plot(kind='bar', color=['green','blue','orange','red'])

plt.title('Average Marks by Subject')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

### Grade Distribution Bar Chart



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the occurrences of each grade
grade_counts = df['Grade'].value_counts().sort_index()

# Define the order of grades for consistent plotting
grade_order = ['A+', 'A', 'B', 'C', 'D', 'F']
grade_counts = grade_counts.reindex(grade_order, fill_value=0)

# Set seaborn style and context for better aesthetics
sns.set_style("whitegrid")
sns.set_context("notebook", font_scale=1.2)

# Create the plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plotting the bar chart using a seaborn palette
# Addressing the FutureWarning by assigning hue and setting legend=False
sns.barplot(x=grade_counts.index, y=grade_counts.values, ax=ax, palette="magma", hue=grade_counts.index, legend=False)

# Add data labels on top of each bar
for i, v in enumerate(grade_counts.values):
    ax.text(i, v + 5, str(v), color='black', ha='center', va='bottom', fontsize=10)

plt.title('Distribution of Grades', fontsize=16, fontweight='bold')
plt.xlabel('Grade', fontsize=14)
plt.ylabel('Number of Students', fontsize=14)
plt.ylim(0, grade_counts.max() * 1.15) # Adjust y-axis limit to accommodate labels
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a light horizontal grid
plt.tight_layout()
plt.show()

sns.reset_orig() # Reset seaborn style to default
```
